### Initial Finding: Delivery Timeliness

Among 96,470 eligible delivered orders, 8.11% arrived after their estimated delivery dates. On-time or early orders had a median delivery time approximately 12.32 days before the estimated date, whereas late orders had a median delay of approximately 5.81 days.

The overall distribution contains extreme early and late observations, which warrant further validation. These findings describe historical delivery performance; they do not yet establish the factors responsible for delays or the predictive ability of a machine learning model.

### Finding: Monthly Delivery Performance

The late-delivery rate varied substantially across order purchase months. It increased from 5.29% in October 2017 to 14.31% in November 2017, reached 21.36% in March 2018, and then fell to 5.31% in April 2018.

These results indicate temporal variation in delivery outcomes. They do not establish the cause of the changes. Further analysis of seller, product-category and geographic patterns is needed. The earliest months also have sparse observations, so their rates should be interpreted cautiously.

### Statistical Analysis: Delivery Timeliness and Customer Satisfaction

The analysis used 95,824 unique eligible orders, with one aggregated review score per order. Late-delivered orders had a mean review score of 2.57, compared with 4.29 for orders delivered on time or early. The mean difference (late minus on-time/early) was −1.7278 points, with a 95% bootstrap confidence interval of [−1.7668, −1.6899].

A two-sided Mann–Whitney U test indicated a statistically significant difference between review-score distributions, with a reported p-value below floating-point display precision. Cliff's delta was −0.5538, indicating a large difference in the direction of lower scores for late-delivered orders.

**Business implication:** Delivery performance and customer satisfaction are strongly associated in this dataset. Improving delivery reliability is a reasonable area for further investigation, but this observational analysis does not establish causation.
